# Power Management

## Linear Voltage Regulator

![Linear Voltage Regulator](images/linear_voltage_regulator.png)

### The Standard Efficiency Formula

$$\eta \approx \frac{V_{out}}{V_{in}} \approx \frac{5V}{12V} \approx 41\%$$

**Why this matters:**
The remaining **59%** of the power is converted entirely into heat.

### Power Dissipation ($P_d$)
To understand why the efficiency is low, you can calculate the power wasted as heat using:

$$P_d = (V_{in} - V_{out}) \cdot I_{out}$$

If this value is high, a large heatsink or a switching regulator (which uses different physical principles to achieve much higher efficiency) is usually required.
